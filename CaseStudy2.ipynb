```python
import pandas as pd

data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}

df = pd.DataFrame(data)

print("Complete Dataset:")
print(df)

print("\nAverage Values:")
print(df[["Study_Hours", "Attendance", "Assignment_Score", "Previous_Score", "Final_Score"]].mean())

highest = df.loc[df["Final_Score"].idxmax()]
print("\nStudent with Highest Final Score:")
print(highest)

lowest = df.loc[df["Final_Score"].idxmin()]
print("\nStudent with Lowest Final Score:")
print(lowest)

print("\nStudents with Attendance Greater Than 80%:")
print(df.loc[df["Attendance"] > 80, "Student"])

print("\nStudents with Final Score Greater Than 80:")
print(df.loc[df["Final_Score"] > 80, "Student"])


# print("\nStudents with Attendance Greater Than 80%:")
# print(df[df["Attendance"] > 80])

# print("\nStudents with Final Score Greater Than 80:")
# print(df[df["Final_Score"] > 80])

print("\nStatistical Summary:")
print(df.describe())
```

TASK-3

```python
import numpy as np

final_scores = np.array(df["Final_Score"])

print("Final Scores:")
print(final_scores)

print("\nMean:", np.mean(final_scores))
print("Median:", np.median(final_scores))
print("Maximum:", np.max(final_scores))
print("Minimum:", np.min(final_scores))
print("Standard Deviation:", np.std(final_scores))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))
plt.scatter(df["Study_Hours"], df["Final_Score"], color="blue")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Study Hours vs Final Score")
plt.grid(True)
plt.show()

plt.figure(figsize=(7, 5))
plt.scatter(df["Attendance"], df["Final_Score"], color="green")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Score")
plt.title("Attendance vs Final Score")
plt.grid(True)
plt.show()

plt.figure(figsize=(7, 5))
plt.hist(df["Final_Score"], bins=5, color="orange", edgecolor="black")
plt.xlabel("Final Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Final Scores")
plt.grid(axis="y")
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Final_Score"], color="purple")
plt.xlabel("Student")
plt.ylabel("Final Score")
plt.title("Student-wise Final Scores")
plt.show()
```

```python
import numpy as np

def L(x):
    return 5*x - 0.2*x**2

x = np.linspace(0, 5, 1000)
y = L(x)

area = np.trapezoid(y, x)

print("Numerical approximation:", area)
```
